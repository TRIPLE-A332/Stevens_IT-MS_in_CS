```python
import numpy as np
```

```python
## Train Data
X = np.array([[1, 3], 
              [2, 2], 
              [3, 1]])
y = [5.5, 10, 27.5]
## Test point
XT = np.array([[2.5, 2.1]])
```

```python
from sklearn.kernel_ridge import KernelRidge
```

```python
klr = KernelRidge(alpha = 1, kernel = 'poly', degree = 3, gamma = 1, coef0 = 0)
klr.fit(X, y)
klr.dual_coef_
```

```python
klr.predict(XT)
```
