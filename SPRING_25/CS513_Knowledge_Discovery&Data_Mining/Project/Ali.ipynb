```python
# Drop unnecessary columns
dataf_cleaned = dataf.drop(columns=['Unnamed: 0', 'ID', 'Customer_ID', 'Name', 'SSN'])

# Convert 'Age' to numeric, coercing errors to NaN
dataf_cleaned['Age'] = pd.to_numeric(dataf_cleaned['Age'], errors='coerce')

# Convert 'Annual_Income' and 'Outstanding_Debt' to numeric, coercing errors to NaN
dataf_cleaned['Annual_Income'] = pd.to_numeric(dataf_cleaned['Annual_Income'], errors='coerce')
dataf_cleaned['Outstanding_Debt'] = pd.to_numeric(dataf_cleaned['Outstanding_Debt'], errors='coerce')

# Handle missing values by filling with mean or mode
for column in dataf_cleaned.select_dtypes(include=['float64', 'int64']).columns:
    dataf_cleaned[column].fillna(dataf_cleaned[column].mean(), inplace=True)

for column in dataf_cleaned.select_dtypes(include=['object']).columns:
    dataf_cleaned[column].fillna(dataf_cleaned[column].mode()[0], inplace=True)

# Verify the cleaned data
print(dataf_cleaned.info())
```
