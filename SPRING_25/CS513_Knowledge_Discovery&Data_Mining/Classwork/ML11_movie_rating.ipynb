```python
import pandas as pd
```

```python
rating_df=pd.read_csv('C:/Users/kdehn/OneDrive - stevens.edu/___Data/CSV_files/Movie_utility_1.csv',index_col='User')
for col in rating_df.select_dtypes(include=['int', 'float']).columns:
    rating_df[col] = rating_df[col].astype(float)



rating_df.head()
```

```python

row_mean=rating_df.mean(numeric_only=True, axis=1) 
print(row_mean)
row_mean_lst=list(row_mean)
print(row_mean_lst)
 
```

```python
df_centered= rating_df.copy()
for i in range(0,len(row_mean_lst)):
    for j in range(0,len(rating_df.columns)):
        if (rating_df.iloc[i,j])> 0:
            adjust=rating_df.iloc[i,j]-row_mean_lst[i]
            # print(adjust)
            df_centered.iloc[i,j]=adjust
df_centered.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cos_sim = pd.DataFrame(cosine_similarity(df_centered))
cos_sim.columns = df_centered.T.columns
cos_sim['User'] = [c for c in df_centered.T.columns]
cos_sim=cos_sim.set_index(['User'])
cos_sim.head()
```
