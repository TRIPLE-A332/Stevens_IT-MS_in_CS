```python
import pandas as pd
from sklearn.metrics import jaccard_score
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv('C:/Users/kdehn/OneDrive - stevens.edu/___Data/CSV_files/movie_actor.csv', index_col='Movie')
data = data.T
data.head()
```

```python
jac_sim = pd.DataFrame(columns = [e for e in data.columns])
jac_sim['index'] = [e for e in data.columns]
jac_sim = jac_sim.set_index(['index'])
for ind in [e for e in data.columns]:
    for col in [e for e in data.columns]:
        jac_sim.loc[ind,col] = jaccard_score(data[ind],data[col])
jac_sim
```

```python
cos_sim = pd.DataFrame(cosine_similarity(data.T))
cos_sim.columns = data.columns
cos_sim['Recipe'] = [e for e in data.columns]
cos_sim = cos_sim.set_index(['Recipe'])
cos_sim
```
