```python
import pandas as pd
import numpy as np
from sklearn.metrics import jaccard_score
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv('D:/Stevens GIT/Stevens_IT-MS_in_CS/SPRING_25/CS513_Knowledge_Discovery&Data_Mining/HW/Recipe.csv')
```

```python
df.head(7)
```

```python
X = df.drop(columns=['Recipe'])
```

```python
# Initialize Jaccard similarity DataFrame
ingredients = X.columns
jac_sim = pd.DataFrame(index=ingredients, columns=ingredients)

# Compute Jaccard similarity between each pair of ingredients
for ind in ingredients:
    for col in ingredients:
        jac_sim.loc[ind, col] = jaccard_score(X[ind], X[col])

# Convert to float
jac_sim = jac_sim.astype(float)

# Display result
jac_sim.head()
```

```python
cosine_sim_matrix = cosine_similarity(X)

# Convert to DataFrame
cosine_df = pd.DataFrame(cosine_sim_matrix, index=df['Recipe'], columns=df['Recipe'])

print("\nCosine Similarity Matrix:\n")
cosine_df.head(7)
```
